```python
import tkinter as tk
from tkinter import filedialog
from pdf2docx import Converter
import os
```

# 2. Convert PDF to Word

```python
# 1. Open a native file picker to select the PDF
# (We hide the main tkinter background window so only the file picker shows)
root = tk.Tk()
root.withdraw()
root.attributes('-topmost', True) 

print("Please select a PDF from the popup window...")
pdf_file = filedialog.askopenfilename(
    title="Select a PDF to convert",
    filetypes=[("PDF Files", "*.pdf")]
)

if not pdf_file:
    print("No file was selected. Please run the cell again.")
else:
    # Create the output filename in the exact same folder
    docx_file = pdf_file[:-4] + '.docx'
    
    # 2. Convert the File
    print(f"\nConverting: {os.path.basename(pdf_file)}")
    print("Please wait...")
    
    cv = Converter(pdf_file)
    cv.convert(docx_file)      
    cv.close()
    
    # 3. Output
    print(f"\n✅ Conversion complete!")
    print(f"Your Word document has been saved directly to:\n{docx_file}")
```
