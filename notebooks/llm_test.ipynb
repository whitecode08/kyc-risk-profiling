```python
import os
from google import genai

client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))

# Gunakan model yang terbukti ada dari hasil list() di atas, misalnya gemini-2.0-flash
response = client.models.generate_content(
    model="gemini-2.0-flash", 
    contents='Tell me a one-sentence joke about programming.',
)

print(response.text)
```
